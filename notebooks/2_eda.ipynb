```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
events_cleanded = pd.read_csv("../data/notebooks/events_cleaned.csv")
```

```python
df_fighters_red = events_cleanded[['fighter_red', 'stance_red']].rename(
    columns={'fighter_red': 'fighter', 'stance_red': 'stance'}
)
df_fighters_blue = events_cleanded[['fighter_blue', 'stance_blue']].rename(
    columns={'fighter_blue': 'fighter', 'stance_blue': 'stance'}
)

# Combine into one fighter dataframe
df_fighters = pd.concat([df_fighters_red, df_fighters_blue], ignore_index=True)

# Drop duplicate fighters (keep their first known stance)
df_fighters = df_fighters.drop_duplicates(subset=['fighter'], keep='first')

# Count unique stances across fighters
stance_counts = df_fighters['stance'].value_counts(dropna=True)

print("Unique stance counts across fighters:\n", stance_counts)
```

```python
# Pick column and count values
counts = events_cleanded['weight_class'].value_counts() 

# Plot
plt.figure(figsize=(12,6))
counts.plot(kind='bar')

plt.title("Weight Class Frequencies", fontsize=16)
plt.xlabel("Weight Classes", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# Pick column and count values
counts = events_cleanded['win_method'].value_counts()

# Plot
plt.figure(figsize=(12,6))
counts.plot(kind='bar')

plt.title("Win Method Frequencies", fontsize=16)
plt.xlabel("Win Methods", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# Plot
plt.figure(figsize=(12,6))
stance_counts.plot(kind='bar')

plt.title("Stance Frequencies", fontsize=16)
plt.xlabel("Stances", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
sns.pairplot(events_cleanded, diag_kind="kde", corner=True)
plt.show()
```

```python
# Compute correlation matrix
# Select only numeric columns
numeric_df = events_cleanded.select_dtypes(include=['int64', 'float64'])
corr = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(24, 18))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", cbar=True, square=True)

plt.title("Correlation Heatmap of Numerical Features", fontsize=16)
plt.tight_layout()
plt.show()
```
